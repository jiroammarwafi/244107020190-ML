### Tujuan Praktikum:
Menerapkan FAISS untuk mencari kesamaan (*similarity query*) pada ruang fitur numerik multi-dimensi setelah distandarisasi menggunakan `StandardScaler`.

```python
import numpy as np
import pandas as pd
import faiss
from sklearn.preprocessing import StandardScaler

# Membuat dataset fitur audio numerik sintetis representatif
np.random.seed(42)
features = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 'instrumentalness', 'tempo']
data_sim = np.random.randn(5000, len(features)).astype('float32')
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_sim).astype('float32')

index_spotify = faiss.IndexFlatL2(len(features))
index_spotify.add(data_scaled)

query_song = data_scaled[0:1]
k = 5
distances, indices = index_spotify.search(query_song, k)
print(f'Indeks lagu dengan kemiripan fitur tertinggi terhadap lagu #0:')
for rank, idx in enumerate(indices[0]):
    print(f'Peringkat {rank+1}: Lagu #{idx} (Jarak L2: {distances[0][rank]:.4f})')
```
