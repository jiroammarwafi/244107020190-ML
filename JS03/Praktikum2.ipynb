2. Melakukan seleksi fitur untuk menyaring atribut-atribut prediktor utama.

## Langkah 1 - Load Data

```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

data_path = r'..\Dataset\Titanic-Dataset.csv'

df = pd.read_csv(data_path)
df.head()
```

## Langkah 2 - Encoding Fitur 'Sex' (Binary Mapping)

```python
# Encoding 'Sex': male -> 1, female -> 0
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})
print('=== Hasil Setelah Encoding Sex ===')
df[['Sex']].head()
```

## Langkah 3 - Encoding Fitur 'Cabin' (Deck Extraction & Imputasi)

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
df['Cabin'] = df['Cabin'].apply(lambda x: x[0] if x != 'Unknown' else 'U')
print('=== Hasil Setelah Encoding Deck Kabin ===')
df[['Cabin']].head(10)
```

## Langkah 4 - Imputasi Fitur 'Age' dengan Median

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
print('=== Hasil Setelah Imputasi Usia (Median) ===')
df[['Age']].head(10)
```

## Langkah 5 - Standardisasi Fitur 'Age'

```python
scaler = StandardScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
print('=== Hasil Setelah Standardisasi Age ===')
df[['Age']].head(10)
```
