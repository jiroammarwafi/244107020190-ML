### Tujuan Praktikum:
1. Memuat citra digital menggunakan pustaka `PIL` (Pillow).
2. Memisahkan kanal warna Red, Green, Blue (RGB).
3. Mengekstrak dan memvisualisasikan histogram intensitas warna sebagai fitur numerik citra.

## Langkah 0 - Persiapan Library

```python
import os
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

## Langkah 1 - Load Gambar

```python
img_path = r'..\Dataset\Lenna.png'

img = Image.open(img_path)
print(f'Format: {img.format}, Dimensi: {img.size}, Mode: {img.mode}')

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title('Citra Asli Lenna')
plt.axis('off')
plt.show()
```

## Langkah 2 - Ekstraksi Fitur Histogram Kanal RGB

```python
# Ekstrak setiap kanal R, G, B
r, g, b = img.split()

print(f'Panjang array fitur histogram kanal Red: {len(r.histogram())} bins')
print('Sampel 20 nilai pertama histogram Red:\n', r.histogram()[:20])
```

## Langkah 3 - Visualisasi Histogram Fitur Warna

```python
plt.figure(figsize=(10, 5))
plt.plot(r.histogram(), color='red', label='Red Channel', alpha=0.8)
plt.plot(g.histogram(), color='green', label='Green Channel', alpha=0.8)
plt.plot(b.histogram(), color='blue', label='Blue Channel', alpha=0.8)
plt.title('Fitur Histogram Distribusi Warna Citra Lenna')
plt.xlabel('Nilai Intensitas Piksel (0 - 255)')
plt.ylabel('Frekuensi Kemunculan')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
