# Bagian 1
## Kegunaan Library
1. PyPREP -> Library berbasis Python yang dirancang untuk melakukan preprocessing (prapemrosesan) data EEG (Electroencephalography). Library ini merupakan implementasi dari PREP pipeline yang populer, berfungsi untuk membersihkan sinyal otak dari artefak, melakukan interpolasi channel yang buruk (bad channels), re-referencing, serta pembersihan derau (noise) secara otomatis.
2. Scipy (Scientific Python) -> Library fundamental untuk komputasi saintifik dan teknis di Python. Scipy menyediakan modul untuk optimasi, aljabar linear, integrasi, interpolasi, pemrosesan sinyal dan gambar (scipy.signal), serta statistik yang sangat sering digunakan dalam pengolahan data sinyal biomedis maupun machine learning.
3. Wandb (Weights & Biases) -> Alat (tool) untuk MLOps yang digunakan untuk melakukan pelacakan eksperimen (experiment tracking), visualisasi metrik pelatihan model (seperti akurasi dan loss secara real-time), manajemen dataset, serta pencatatan artefak model machine learning atau deep learning.
4. pyECG: Library Python open-source yang digunakan untuk membaca, menulis, dan mengelola format dataset sinyal jantung atau ECG (Electrocardiogram) yang umum, seperti format Physiobank WFDB (.hea/.dat), sehingga memudahkan peneliti mengakses dan mengekstrak sinyal elektrik jantung.
