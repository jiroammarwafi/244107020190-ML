# Praktikum 3 - (Sekilas) Seleksi Fitur, Encoding, dan Standardisasi

## Langkah 0 - Persiapan Library
Mengimpor library yang dibutuhkan untuk manipulasi data dan pra-pengolahan.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Langkah 1 - Load Data
Memuat dataset Titanic yang telah diperbaiki pada tahap imputasi sebelumnya (`Titanic-Dataset-fixed.csv`).

```python
data_path = r'..\Dataset\Titanic-Dataset-fixed.csv'
df = pd.read_csv(data_path)
df.head()
```

## Langkah 2 - Slice Data (Seleksi Fitur)
Memilih fitur yang relevan untuk memprediksi keselamatan penumpang:
- `Survived` (Label/Target)
- `Pclass` (Fitur Ordinal: Kelas Penumpang)
- `Age` (Fitur Numerik: Umur)
- `Sex` (Fitur Kategorikal: Jenis Kelamin)
- `Cabin` (Fitur Kategorikal: Nomor Kabin/Dek)

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

## Langkah 3 - Encoding
Melakukan transformasi data kategorikal non-numerik (`Sex` dan `Cabin`) menjadi representasi angka menggunakan `LabelEncoder`.

```python
le = LabelEncoder()
# Encoding kolom Sex
df['Sex'] = le.fit_transform(df['Sex'])
# Encoding kolom Cabin
df['Cabin'] = le.fit_transform(df['Cabin'])
```

## Langkah 4 - Verifikasi Hasil Encoding
Memeriksa 5 baris pertama data setelah proses encoding.

```python
df.head()
```

## Langkah 5 - Standarisasi
Melakukan standardisasi (z-score scaling) pada fitur `Age` agar berdistribusi normal standar dengan mean ~ 0 dan standar deviasi ~ 1.

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```

## Langkah 6 - Verifikasi Hasil Standarisasi
Memeriksa kembali hasil dataset setelah fitur `Age` distandarisasi.

```python
df.head()
```
