# Praktikum 4 - Pra Pengolahan Data: Data Tidak Terstruktur (Citra)

## Langkah 0 - Persiapan Library
Mengimpor OpenCV untuk manipulasi citra dan Matplotlib untuk visualisasi.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Langkah 1 - Load Gambar
Memuat citra `Lenna.png`. Secara default, OpenCV membaca citra dalam susunan kanal **BGR** (Blue, Green, Red). Oleh karena itu, kita perlu mengonversinya menjadi **RGB** agar warna tampil natural ketika divisualisasikan dengan Matplotlib.

```python
img_path = r'..\Dataset\Lenna.png'
# Membaca citra (format BGR)
img = cv2.imread(img_path)

# Konversi BGR -> RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tampilkan informasi dimensi asli
print(f'Dimensi citra asli (Tinggi, Lebar, Kanal): {img_rgb.shape}')

# Visualisasi citra asli
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title('Gambar Asli (Format RGB)')
plt.axis('off')
plt.show()
```

## Langkah 2 - Resize Gambar
Mengubah resolusi spasial citra menjadi 128x128 piksel untuk standardisasi ukuran input model pembelajaran mesin.

```python
# Resize citra ke ukuran 128x128 piksel
img_resized = cv2.resize(img_rgb, (128, 128))

print(f'Dimensi setelah resize: {img_resized.shape}')

# Visualisasi hasil resize
plt.figure(figsize=(5, 5))
plt.imshow(img_resized)
plt.title('Gambar setelah Resize (128x128 px)')
plt.axis('off')
plt.show()
```

## Langkah 3 - Konversi ke Grayscale
Mengonversi citra warna 3-kanal (RGB) menjadi citra derajat keabuan (grayscale 1-kanal) untuk mereduksi kompleksitas komputasi.

```python
# Konversi citra resize (RGB) ke Grayscale
img_gray = cv2.cvtColor(img_resized, cv2.COLOR_RGB2GRAY)

print(f'Dimensi citra grayscale: {img_gray.shape}')

# Visualisasi citra grayscale
plt.figure(figsize=(5, 5))
plt.imshow(img_gray, cmap='gray')
plt.title('Gambar Grayscale (128x128 px, 1 Kanal)')
plt.axis('off')
plt.show()
```

## Perbandingan Visual Seluruh Tahapan Pra Pengolahan Citra
Menampilkan ketiga tahapan secara berdampingan untuk melihat perbedaan detail dan informasi visual.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img_rgb)
axes[0].set_title(f'Asli {img_rgb.shape[:2]}')
axes[0].axis('off')

axes[1].imshow(img_resized)
axes[1].set_title(f'Resize {img_resized.shape[:2]}')
axes[1].axis('off')

axes[2].imshow(img_gray, cmap='gray')
axes[2].set_title(f'Grayscale {img_gray.shape}')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
